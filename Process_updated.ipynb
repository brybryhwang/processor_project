## FSM simulation

Demonstrate that the **ControlFSM** performs the correct state
transitions for each processor opcode.  

```python
from graphviz import Digraph

g = Digraph('FSM', format='png', graph_attr={'rankdir':'LR'})
states = ['FETCH','DECODE','EXEC_R','ADDR','MEM_RD','MEM_WR',
          'WB_ALU','WB_MEM','BRANCH','JUMP','LDPC']
for s in states:
    g.node(s)

g.edge('FETCH', 'DECODE')
g.edge('DECODE', 'EXEC_R', label='R-type')
g.edge('DECODE', 'ADDR',   label='LOAD / STORE')
g.edge('DECODE', 'BRANCH', label='BEQ')
g.edge('DECODE', 'JUMP',   label='JMP')
g.edge('DECODE', 'LDPC',   label='LDPC')
g.edge('EXEC_R','WB_ALU')
g.edge('ADDR',  'MEM_RD')
g.edge('ADDR',  'MEM_WR')
g.edge('MEM_RD','WB_MEM')
# final returns
for s in ['WB_ALU','WB_MEM','MEM_WR','BRANCH','JUMP','LDPC']:
    g.edge(s, 'FETCH')

g
```
